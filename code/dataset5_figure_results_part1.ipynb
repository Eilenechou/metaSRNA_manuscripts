```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm

fm.fontManager.addfont(str("arial.ttf"))

arial_font = fm.FontProperties(fname="arial.ttf", size=20)

arial_font_in = fm.FontProperties(fname="arial.ttf", size=16)

pairs = [61.1063, 66.2247, 26.4408, 39.3013, 59.4065, 71.1567, 43.1119, 81.624, 50.997, 77.5031]              # values for the pair bars
singles = [67.8982,93.9295,84.5536,81.1142]      # values for the 4 single bars
# sample_a = [61.1063, 26.4408, 59.4065, 43.1119, 50.997]
# sample_b = [66.2247,39.3013,71.1567,81.624,77.5031]
# sample_c = [67.8982,93.9295,84.5536,81.1142]

x = np.arange(5 + len(singles))  # 0 = pair group, 1..4 = single bars

fig, ax = plt.subplots(figsize=(8,7))


bar_width = 0.35
for i in range(5):
    bar1 = ax.bar(x[i] - bar_width/2, pairs[2*i], width=bar_width, color="#1E90FF", label="Compared paper" if i == 0 else "_nolegend_")
    bar2 = ax.bar(x[i] + bar_width/2, pairs[2*i+1], width=bar_width, color="red", label="metaSRNA" if i == 0 else "_nolegend_")
    for bar in bar1 + bar2:   
        height = bar.get_height()
        if height > 0:
            ax.text(
                bar.get_x() + bar.get_width()/2, height + 0.2,
                f"{int(height)}",   
                ha="center", va="bottom", rotation=45, fontproperties=arial_font_in
            )


for i, val in enumerate(singles, start=5):
    ax.bar(x[i], val, width=0.6, color="gray", label="both" if i == 5 else "_nolegend_")
    ax.text(
        x[i], val + 0.2,  
        str(int(val)),rotation=45,    
        ha='center',fontproperties=arial_font_in 
    )

ax.set_xticks(x)
ax.set_xticklabels(
    ["SRR8195524", "SRR8195525", "SRR8195527", "SRR8195530", "SRR8195532",
    "SRR8195526","SRR8195528","SRR8195529","SRR8195531"], rotation=45, ha='right',fontproperties=arial_font
)
ax.set_ylabel("Samples")      
ax.set_ylabel('Mapping percentage(%)',fontproperties=arial_font)
ax.set_title('Comparison of mapping percentages between paper and our Genome selection',fontproperties=arial_font)
ax.set_ylim(0, 100)
ax.legend()

plt.xticks(fontproperties=arial_font)
plt.yticks(fontproperties=arial_font)
plt.tight_layout()
# plt.grid(True)
# plt.grid(True, color="lightgray", linewidth=0.1)
ax.legend(prop=arial_font, title="Groups",  bbox_to_anchor=(1.02, 1), title_fontproperties=arial_font)

# plt.show()
plt.savefig("figure3_B.pdf", format="pdf", bbox_inches="tight")


```
