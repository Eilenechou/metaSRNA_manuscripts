```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib import font_manager as fm

fm.fontManager.addfont("arial.ttf")
arial_font = fm.FontProperties(fname="arial.ttf", size=30)
arial_font_in = fm.FontProperties(fname="arial.ttf", size=30)

total_names = ["SRR18078872", "SRR18078871", "SRR18078870","SRR18078869", "SRR18078868", "SRR18078867"]
group_name = ["Control 1", "Control 2", "Control 3", "DSS 1", "DSS 2", "DSS 3"]

color_map = {
    'Roseovarius nitratireducens': 'gray',

    'Ectobacillus ponti': 'chocolate',
    'Microbacterium karelineae': 'tomato',
    'Xanthomonas sontii': 'yellowgreen'

}
unique_species = list(color_map.keys())

group_species = {
    "Control 1": ['Roseovarius nitratireducens'],
    "Control 2": ['Roseovarius nitratireducens'],
    "Control 3": ['Roseovarius nitratireducens','Microbacterium karelineae','Xanthomonas sontii','Ectobacillus ponti'],
    "DSS 1": ['Microbacterium karelineae', 'Xanthomonas sontii'],
    "DSS 2": ['Microbacterium karelineae'],
    "DSS 3": []
}

fig, ax = plt.subplots(figsize=(15, 6))
bar_width = 0.8

for i, group in enumerate(group_name):
    species_list = group_species[group]
    if len(species_list) == 0:
        rect = plt.Rectangle(
            (i - bar_width/2, 0), bar_width, 1.0,
            facecolor='white',
            edgecolor='black',
            linestyle=':',
            linewidth=2
        )
        ax.add_patch(rect)
        # ax.text(i, 0.5,  ha="center", va="center",
        #         fontproperties=arial_font_in, color="black")
        continue
    n = len(species_list)
    if n > 0:
        height_per_species = 1.0 / n
    else:
        height_per_species = 1
    bottom = 0

    for sp in species_list:
        color = color_map[sp]
        ax.bar(i, height_per_species, bottom=bottom, width=bar_width,
               color=color, edgecolor="black", linewidth=0.3)
        bottom += height_per_species

ax.set_xticks(range(len(total_names)))
ax.set_xticklabels([f"{group_name[i]}" for i, name in enumerate(total_names)], rotation=45, ha='right' ,fontproperties=arial_font)
ax.set_ylabel("Relative Abundance", fontproperties=arial_font)
ax.set_ylim(0, 1)

legend_patches = [Patch(facecolor=color_map[sp], label=sp) for sp in unique_species]
legend_patches.append(Patch(facecolor='white', edgecolor='black', linestyle=':', label='No data'))

ax.legend(
    handles=legend_patches,
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),
    ncol=1,
    prop=arial_font_in,
    title="Species",
    title_fontproperties=arial_font_in
)

# ax.spines['top'].set_visible(False)
# ax.spines['right'].set_visible(False)
# ax.spines['left'].set_visible(False)
# ax.tick_params(left=False)


plt.xticks(fontproperties=arial_font_in)
plt.yticks(fontproperties=arial_font_in)
plt.tight_layout()
plt.savefig("figure2_G2.pdf", format="pdf", bbox_inches="tight")
```
