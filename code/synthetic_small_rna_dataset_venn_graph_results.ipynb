```python
import matplotlib.pyplot as plt
from matplotlib import font_manager
font_manager.fontManager.addfont("arial.ttf")
plt.rcParams["font.family"] = "Arial"
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 20,
    "pdf.fonttype": 42,
})
plt.rcParams["hatch.linewidth"] = 1
```

```python
import matplotlib.pyplot as plt
# from matplotlib_venn import venn2
from matplotlib_venn import venn2_unweighted
#mapping:8931002/total:10879366
plt.figure(figsize=(6, 6))
venn = venn2_unweighted(subsets=("0","1948364\n(17.91%)","977\n(82.09%)")
, set_labels=('published_977_sequences', 'SRR5234257(4N_A)\nsequences length (16-25)'),set_colors=("red", "green"))
for label in venn.subset_labels:
    if label is not None:
        label.set_fontsize(20)
# venn.get_patch_by_id('10').set_facecolor("#A5CFEF")
# venn.get_patch_by_id('01').set_facecolor("#C7AEDB")
# venn.get_patch_by_id('11').set_facecolor("#E7A6A6")
plt.title("Venn Diagram:\n Overlap between published synthetic small RNA dataset (n = 977)\n and our method-detected sequences in SRR5234257: 4N_A")
plt.savefig("supp_figure1_4N_A.pdf", format="pdf", bbox_inches="tight")
```
