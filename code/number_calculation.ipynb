```python
from Bio import SeqIO

min_len = 18
valid = 0

for record in SeqIO.parse("final_seq_12.fasta", "fasta"):
    if len(record.seq) >= min_len and len(record.seq) <= 40:
        valid += 1

print(f"Sequences ≥ {min_len} bp:", valid)
```
