```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df_space_axis = pd.read_csv("hairpin_data/hairpin_embedding_output.csv", sep=",",header=None)
print(df_space_axis)
# print(df_space_axis.dtypes)


scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_space_axis)


pca = PCA(n_components=3)
principal_components = pca.fit_transform(df_scaled)


pca_df = pd.DataFrame(principal_components, columns=['X', 'Y', 'Z'])

pca_df.to_csv("hairpin_data/axis_results.csv", index=False)
# df_2 = pd.read_csv("data/whole_paper_sequences.csv", sep="\t")
# # Apply the function to df_1['sequence']
# df_1['original_sequence'] = df_1['sequence'].apply(is_near_substring)
```
